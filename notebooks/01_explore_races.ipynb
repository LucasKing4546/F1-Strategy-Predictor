# Explore one 2025 and one 2026 race
Hungarian GP, FP2 and race.

```python
import logging

import fastf1
import fastf1.plotting
import matplotlib.pyplot as plt
import pandas as pd

from f1strat.data.cache import enable_cache

enable_cache()
fastf1.set_log_level(logging.WARNING)
fastf1.plotting.setup_mpl(mpl_timedelta_support=False, color_scheme=None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
```

```python
EVENT = "Hungarian"
sessions = {}
for year in (2025, 2026):
    for kind in ("FP2", "R"):
        session = fastf1.get_session(year, EVENT, kind)
        session.load(telemetry=False)
        sessions[f"{year} {kind}"] = session
```

## Laps table

```python
for key, session in sessions.items():
    laps = session.laps
    print(f"{key}: {len(laps)} laps, {laps.Driver.nunique()} drivers, {session.total_laps} race laps")
    display(laps.head())
```

## Lap time vs lap, per driver, coloured by compound

```python
def plot_lap_times(session, ncols=4):
    laps = session.laps.assign(LapTimeS=session.laps.LapTime.dt.total_seconds()).dropna(subset=["LapTimeS"])
    drivers = [d for d in session.results.Abbreviation if d in set(laps.Driver)]
    colors = fastf1.plotting.get_compound_mapping(session=session)
    nrows = -(-len(drivers) // ncols)
    fig, axes = plt.subplots(nrows, ncols, figsize=(4 * ncols, 2.4 * nrows), sharex=True, sharey=True)
    for ax, driver in zip(axes.flat, drivers):
        d = laps[laps.Driver == driver]
        ax.scatter(d.LapNumber, d.LapTimeS, c=d.Compound.map(colors).fillna("grey"), s=10, edgecolors="black", linewidths=0.3)
        ax.set_title(driver, fontsize=9)
    for ax in axes.flat[len(drivers):]:
        ax.set_visible(False)
    best = laps.LapTimeS.min()
    axes.flat[0].set_ylim(best - 1, best * 1.15)
    used = set(laps.Compound.dropna())
    handles = [plt.Line2D([], [], marker="o", ls="", color=c, markeredgecolor="black", label=k) for k, c in colors.items() if k in used]
    fig.legend(handles=handles, loc="upper right", ncols=len(handles))
    fig.suptitle(f"{session.event.year} {session.event.EventName} {session.name}")
    fig.supxlabel("Lap")
    fig.supylabel("Lap time (s)")
    fig.tight_layout(rect=(0, 0, 1, 0.97))
    plt.show()
```

```python
plot_lap_times(sessions["2025 FP2"])
```

```python
plot_lap_times(sessions["2025 R"])
```

```python
plot_lap_times(sessions["2026 FP2"])
```

```python
plot_lap_times(sessions["2026 R"])
```
